```python
#Case 1: Assume the mean intensity of the image as the threshold value.
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\saraj\Downloads\baby_image.jpg", 0)

threshold = np.mean(img)
print("Mean Threshold Value =", threshold)
_, binary = cv2.threshold(img, threshold, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis("off")
plt.show()
```

```python
# Case 2: User Inputs Threshold
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\saraj\Downloads\baby_image.jpg", 0)

threshold = int(input("Enter threshold value (0-255): "))

_, binary = cv2.threshold(img, threshold, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis("off")
plt.show()
```

```python
#2. Write a program (WAP) to convert an RGB (color) image into a grayscale image.
#Case 1: Convert the RGB image to grayscale by taking the mean average of the three color planes (R, G, and B).
#Case 2: Input the weightages of the three color planes (R, G, and B) from the user. The weightage of each plane should be a value between 0 and 1, and the sum of all three weightages must be equal to 1.
#Example:
#Weightages of R, G, and B are 0.7, 0.2, and 0.1, respectively.

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\saraj\Downloads\baby_image.jpg")

# Convert BGR to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]
# mean formula
gray = ((R + G + B) / 3).astype(np.uint8)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original RGB Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale (Mean Method)")
plt.axis("off")
plt.show()

cv2.imwrite("gray_mean.jpg", gray)
```
