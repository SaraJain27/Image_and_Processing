```python
# Write a program to perform the following operations on images and videos:
#a.	Read, display, and write an image.
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread(r"C:\Users\saraj\Downloads\baby_image.jpg")

# Check image loaded or not
if img is None:
    print("Image not found!")
else:
    # Convert BGR to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Display image
    plt.imshow(img_rgb)
    plt.axis("off")
    plt.show()

    # Save image
    cv2.imwrite("output.jpg", img)
    print("Image saved successfully!")
```

```python
# c.	 Perform image resizing.

import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\saraj\Downloads\baby_image.jpg")

# Resize image
resized = cv2.resize(img, (400, 300))

# Convert to RGB for display
resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

plt.imshow(resized_rgb)
plt.axis("off")
plt.show()

cv2.imwrite("baby_resized.jpg", resized)
```

```python
# d.	 Convert a colored image into a grayscale image.
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\saraj\Downloads\baby_image.jpg")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap='gray')
plt.axis("off")
plt.show()

cv2.imwrite("baby_gray.jpg", gray)
```

```python
# e.	 Read and display a video.
import cv2

# Load video
video = cv2.VideoCapture("video.mp4")

while True:
    ret, frame = video.read()

    # Stop when video ends
    if not ret:
        break

    cv2.imshow("Video Player", frame)

    # Press q to quit
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```

```python
# f) Extract Frames from a Video
import cv2
import os

video = cv2.VideoCapture("video.mp4")

# Folder for frames
os.makedirs("frames", exist_ok=True)

count = 0

while True:
    ret, frame = video.read()

    if not ret:
        break

    cv2.imwrite(f"frames/frame_{count}.jpg", frame)

    count += 1

video.release()

print("Total Frames Extracted:", count)
```

```python
import cv2
import os

video = cv2.VideoCapture("video.mp4")

fps = int(video.get(cv2.CAP_PROP_FPS))

os.makedirs("frames", exist_ok=True)

count = 0
frame_no = 0

while True:
    ret, frame = video.read()

    if not ret:
        break

    # Save one frame every second
    if frame_no % fps == 0:
        cv2.imwrite(f"frames/frame_{count}.jpg", frame)
        count += 1

    frame_no += 1

video.release()

print("Frames saved:", count)
```
